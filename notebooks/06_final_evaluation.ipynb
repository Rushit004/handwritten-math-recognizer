# 06. Final Evaluation, Confusion Matrix & Error Analysis

Evaluating the best performing model (Lightweight 2D CNN, 94.57% test accuracy) on the untouched 5,745 test samples.


```python
import json
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

with open('../results/confusion_matrix/top_confused_pairs.json') as f:
    pairs = json.load(f)

print("Top Confused Symbol Pairs:")
for p in pairs[:5]:
    print(f"  True: {p['true_class']} ({p['true_latex']}) -> Pred: {p['pred_class']} ({p['pred_latex']}) | {p['error_count']} misclassifications")
```
