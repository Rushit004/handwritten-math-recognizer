# 03. Feature Engineering: Raw Pixels vs HOG Descriptors

Comparing:
- **Representation A:** Flattened raw pixel intensities (32x32 = 1,024 features)
- **Representation B:** Histogram of Oriented Gradients (HOG) with 9 orientations, 8x8 cells, 2x2 blocks (324 features)

HOG captures local edge directions and gradient magnitudes, providing translation and minor stroke-thickness invariance.


```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from src.feature_engineering import FeaturePipeline, get_hog_visualization
from src.data_preprocessing import load_processed_dataset

X_train, y_train, _, _, _, _, _, _, _ = load_processed_dataset('../data/processed/math_symbols_dataset.npz')
feat_pipe = FeaturePipeline.load('../models/feature_pipeline.joblib')

raw_feats = feat_pipe.transform(X_train[:100], feature_type='raw')
hog_feats = feat_pipe.transform(X_train[:100], feature_type='hog')

print(f"Raw pixel representation dimension: {raw_feats.shape[1]}")
print(f"HOG feature representation dimension: {hog_feats.shape[1]}")
```
