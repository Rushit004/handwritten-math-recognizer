# 04. Baseline Model Development & Feature Benchmark

We establish a rigorous baseline reference using L2-regularized Logistic Regression and directly measure the empirical gain of HOG feature engineering over Raw Pixels.


```python
import json
import pandas as pd

with open('../results/metrics/metrics_summary.json') as f:
    summary = json.load(f)

df = pd.DataFrame(summary).T
print("Test Set Performance Comparison:")
df.loc[['Baseline (LogReg Raw)', 'Baseline (LogReg HOG)'], ['accuracy', 'balanced_accuracy', 'macro_f1', 'inference_latency_ms']]
```

### Key Scientific Observation:
HOG features boost accuracy from **81.03% to 92.90%** (+11.87%) and Macro F1 from **76.90% to 88.63%** (+11.73%), while cutting inference latency by more than half. This experimentally proves the necessity of structural geometric descriptors over raw photometric pixels.
